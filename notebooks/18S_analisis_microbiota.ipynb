```python
# Crear un Df con los datos de las columna: 'Level_3', 'CV:mean rel. freq.(%)', HecesP6:rel. freq.(%), 'HecesP7:rel. freq.(%)', 'HecesP8:rel. freq.(%)', 'HecesP9:rel. freq.(%)', 'HecesP10:rel. freq.(%)', 'HecesP11:rel. freq.(%)', 'HecesP12:rel. freq.(%)', 'HecesP13:rel. freq.(%)', 'HecesP14:rel. freq.(%)'

df_cvh = df_cvh[['p-values','Level_3', 'CV: mean rel. freq. (%)', 'HecesP6: rel. freq. (%)', 'HecesP7: rel. freq. (%)', 
          'HecesP8: rel. freq. (%)', 'HecesP9: rel. freq. (%)', 'HecesP10: rel. freq. (%)', 
          'HecesP11: rel. freq. (%)', 'HecesP12: rel. freq. (%)', 'HecesP13: rel. freq. (%)', 
          'HecesP14: rel. freq. (%)']]

print(df_cvh.info())
display(df_cvh.head())
```

#### Cambiamos nombres de las comlumnas

```python
# Cambiar los nombres de las columnas: CV: mean rel. freq. (%) por C. virginica, HecesP6: rel. freq. (%) por F1, HecesP7: rel. freq. (%) por F2, HecesP8: rel. freq. (%) por F3, HecesP9: rel. freq. (%) por F4, HecesP10: rel. freq. (%) por F5, HecesP11: rel. freq. (%) por F6, HecesP12: rel. freq. (%) por F7, HecesP13: rel. freq. (%) por F8, HecesP14: rel. freq. (%) por F9
df_cvh.rename(columns={'Level_3': 'Filo',
                       'CV: mean rel. freq. (%)': 'C. virginica',
                       'HecesP6: rel. freq. (%)': 'F1',
                       'HecesP7: rel. freq. (%)': 'F2',
                       'HecesP8: rel. freq. (%)': 'F3',
                       'HecesP9: rel. freq. (%)': 'F4',
                       'HecesP10: rel. freq. (%)': 'F5',
                       'HecesP11: rel. freq. (%)': 'F6',
                       'HecesP12: rel. freq. (%)': 'F7',
                       'HecesP13: rel. freq. (%)': 'F8',
                       'HecesP14: rel. freq. (%)': 'F9'}, inplace=True)

print(df_cvh.info())
display(df_cvh.head()) 
```

```python
# Ordenar los valores mas altos esten arriba C. virginica F1, F2, F3, F4, F5, F6 y F7
df_cvh.sort_values(by='p-values', ascending=True)
df_cvh = df_cvh.drop(columns=['p-values'])
display(df_cvh.head())

```

```python
# Eliminar filas con cualquier tipo de "classified" al final
df_cvh = df_cvh[~df_cvh['Filo'].str.contains(r'classified$', regex=True)]
```

```python
# Filtrar filos no deseados
filos_a_eliminar = [
    'unknown_unclassified', 'Archaeplastida', 'Streptophyta',
    'Cryptophyta', 'Chlorophyta', 'Bacillariophyta',
    'Cyanobacteria_Chloroplast', 'Cryptophyta:nucl'
]
df_cvh = df_cvh[~df_cvh['Filo'].isin(filos_a_eliminar)]

# Establecer 'Filo' como índice y transponer
df_cvh.set_index('Filo', inplace=True)
df_cvh_f = df_cvh.T

# Normalizar por muestra (%)
df_cvh_norm_f = df_cvh_f.div(df_cvh_f.sum(axis=1), axis=0) * 100

# Top 10 filos más abundantes globalmente
top10_filos = df_cvh_norm_f.sum().sort_values(ascending=False).head(10).index.tolist()

# Filtrar y calcular "Otros"
df_top10 = df_cvh_norm_f[top10_filos].copy()
df_top10["Otros"] = (100 - df_top10.sum(axis=1)).clip(lower=0)

# Agrupar columnas con nombres repetidos
df_top10_grouped = df_top10.groupby(df_top10.columns, axis=1).sum()

# Colores personalizados
cmap = get_cmap('tab10')
colores_dict = {}
for i, col in enumerate(df_top10_grouped.columns):
    if col == 'Otros':
        colores_dict[col] = 'grey'  # negro para "Otros"
    elif col == 'Tubulinea':
        colores_dict[col] = '#1f3b99'  # azul oscuro para Tubulinea
    else:
        colores_dict[col] = cmap(i % cmap.N)

# Graficar
df_top10_grouped.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6),
    color=[colores_dict[col] for col in df_top10_grouped.columns]
)

# Personalización
plt.ylabel("Relative abundance (%)")
plt.xlabel("Samples")
plt.title("Comparative Phyla Composition in $\mathit{C. virginica}$ and Human Feces")
plt.legend(title="Phylum", bbox_to_anchor=(1.05, 1), loc='upper left', frameon=False)
plt.xticks(rotation=45)
plt.ylim(0, 100)
plt.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.cm import get_cmap
import re

# ----- opcional: ordenar muestras por similitud si hay scipy -----
try:
    from scipy.spatial.distance import pdist
    from scipy.cluster.hierarchy import linkage, leaves_list
    SCIPY_OK = True
except Exception:
    SCIPY_OK = False

# =========================
# 0) Copia y normaliza la forma de 'Filo'
# =========================
df = df_all_18SF.copy()  # Debe contener 'Filo', 'p-values' (si existe) y columnas de muestras

# Si 'Filo' no está como columna, intenta recuperarlo del índice
if 'Filo' not in df.columns:
    if df.index.name == 'Filo':
        df = df.reset_index()
    else:
        df = df.reset_index()
        if 'Filo' not in df.columns:
            if 'index' in df.columns:
                df = df.rename(columns={'index': 'Filo'})
            else:
                raise KeyError("No se encontró la columna 'Filo' ni estaba como índice.")

# Arreglo de etiquetas conocidas
df['Filo'] = df['Filo'].replace({
    'Cryptista:nucl': 'Cryptista',
    'Cryptophyta:nucl': 'Cryptophyta'
})

# =========================
# 1) Filtrado y columnas de muestras
# =========================
filos_a_eliminar = ['Eukaryota_unclassified', 'Streptophyta', 'Cryptophyta:nucl']
df = df[~df['Filo'].isin(filos_a_eliminar)]

# Columnas de abundancia (todas menos 'Filo', 'p-values')
cols_muestras = [c for c in df.columns if c not in ['Filo', 'p-values']]
# Quita 'SumaTotal' si estuviera como "muestra"
if 'SumaTotal' in cols_muestras:
    cols_muestras.remove('SumaTotal')

df[cols_muestras] = df[cols_muestras].apply(pd.to_numeric, errors='coerce').fillna(0)

# =========================
# 2) Agregación por Filo (abundancias y p-values)
# =========================
abund_por_filo = df.groupby('Filo')[cols_muestras].sum()

if 'p-values' in df.columns:
    # Usamos el máximo por Filo para ordenar por p-value descendente
    pval_por_filo = df.groupby('Filo')['p-values'].max()
else:
    pval_por_filo = pd.Series(index=abund_por_filo.index, dtype=float)

# =========================
# 3) Top 20 por p-value (desc)  ← selección y orden definidos por p-value
# =========================
if pval_por_filo.notna().any():
    top20_filos = pval_por_filo.sort_values(ascending=False).head(20).index.tolist()
    orden_por_pval = top20_filos[:]  # ya están por p-value
else:
    # sin p-values, cae al top-20 por abundancia
    top20_filos = abund_por_filo.sum(axis=1).sort_values(ascending=False).head(20).index.tolist()
    orden_por_pval = top20_filos


# =========================
# 4) Normalización correcta a % sobre el TOTAL por muestra
# =========================
# total por muestra considerando TODOS los filos
total_por_muestra = abund_por_filo.sum(axis=0).replace(0, np.nan)

# matriz normalizada completa (todas las columnas Filo) a %
full_norm = (abund_por_filo.T
             .div(total_por_muestra, axis=0)
             .fillna(0) * 100)

# submatriz solo con top20 en el orden pedido
top_norm = full_norm[orden_por_pval].copy()

# =========================
# 5) "Otros" = 100 - suma(top20) por muestra
# =========================
otros_pct = (100 - top_norm.sum(axis=1)).clip(lower=0)
mat_norm = top_norm.copy()
mat_norm['Otros'] = otros_pct

# =========================
# 6) (Opcional) Ordenar muestras por similitud (Bray–Curtis) usando mat_norm
# =========================
if SCIPY_OK and mat_norm.shape[0] > 2:
    try:
        D = pdist(mat_norm.values, metric='braycurtis')
        Z = linkage(D, method='average')
        order_idx = leaves_list(Z)
        mat_norm = mat_norm.iloc[order_idx]
    except Exception:
        pass  # si falla, deja el orden actual

# =========================
# 6.1) Reordenación manual del eje X
# =========================
def mover_despues(lista, item, despues_de):
    """Mueve 'item' justo después de 'despues_de' en la lista."""
    if item in lista and despues_de in lista:
        lista = [x for x in lista if x != item]
        idx = lista.index(despues_de)
        lista.insert(idx + 1, item)
    return lista

def mover_antes(lista, item, antes_de):
    """Mueve 'item' justo antes de 'antes_de' en la lista."""
    if item in lista and antes_de in lista:
        lista = [x for x in lista if x != item]
        idx = lista.index(antes_de)
        lista.insert(idx, item)
    return lista

orden_x = list(mat_norm.index)

# --- Caso 1: mover C. virginica después de F9 ---
aliases_cvir = ["C. virginica", "Crassostrea virginica", "C. virginica (promedio)"]
alias_cvir_presente = next((a for a in aliases_cvir if a in orden_x), None)

aliases_f9 = ["F9", "HecesP14", "F-9"]
alias_f9_presente = next((a for a in aliases_f9 if a in orden_x), None)

if alias_cvir_presente and alias_f9_presente:
    orden_x = mover_despues(orden_x, alias_cvir_presente, alias_f9_presente)

# --- Caso 2: mover RMW antes de GZW ---
alias_rmw = next((a for a in ["RMW"] if a in orden_x), None)
alias_gzw = next((a for a in ["GZW"] if a in orden_x), None)

if alias_rmw and alias_gzw:
    orden_x = mover_antes(orden_x, alias_rmw, alias_gzw)

# Aplicar nuevo orden si cambió
mat_norm = mat_norm.loc[orden_x]


# =========================
# 7) Estilo “journal-ready” y gráfico
# =========================
# Ordena la leyenda por abundancia total (mejor lectura)
totales_leyenda = mat_norm[orden_por_pval].sum(axis=0).sort_values(ascending=False)
cols_plot = totales_leyenda.index.tolist() + ['Otros']

mpl.rcParams.update({
    "figure.dpi": 150, "savefig.dpi": 600, "savefig.bbox": "tight",
    "pdf.fonttype": 42, "ps.fonttype": 42,        # embebe fuentes
    "axes.linewidth": 0.8,
    "axes.labelsize": 9, "xtick.labelsize": 8, "ytick.labelsize": 8,
    "legend.fontsize": 7,
    "font.family": "sans-serif",
    "font.sans-serif": ["Helvetica", "Arial", "DejaVu Sans"],
})

# Paleta colorblind-safe + gris para “Otros”
tol = ["#332288","#88CCEE","#44AA99","#117733","#999933",
       "#DDCC77","#CC6677","#882255","#AA4499","#6699CC"]
tab20 = [plt.cm.tab20(i) for i in range(20)]
palette = (tol + tab20)[:len(cols_plot)-1] + ["#4D4D4D"]
color_map = {g: c for g, c in zip(cols_plot, palette)}

fig, ax = plt.subplots(figsize=(7.09, 3.5))
mat_plot = mat_norm[cols_plot]  # ya está normalizada y con 'Otros'
mat_plot.plot(kind='bar', stacked=True, ax=ax,
              color=[color_map[c] for c in cols_plot],
              width=0.92, edgecolor="none")

# Ejes y grilla
ax.set_ylabel("Abundancia relativa (%)")
ax.set_xlabel("Muestras")
ax.set_ylim(0, 100)
ax.yaxis.set_major_locator(mpl.ticker.MultipleLocator(20))
ax.yaxis.set_minor_locator(mpl.ticker.MultipleLocator(10))
ax.grid(axis="y", which="major", linestyle="--", linewidth=0.5, alpha=0.5)
ax.grid(axis="y", which="minor", linestyle=":",  linewidth=0.35, alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.tick_params(axis="x", rotation=45)

# Leyenda (sin itálicas en Filo)
ax.legend(title="Filo",
          bbox_to_anchor=(1.01, 1), loc='upper left',
          frameon=False, ncol=1, title_fontsize=8)

fig.tight_layout()
fig.savefig("fig18S_filos_top20_pub.tiff", dpi=600,
            pil_kwargs={'compression': 'tiff_lzw'})
fig.savefig("fig18S_filos_top20_pub.pdf")
plt.show()

```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import re
from textwrap import fill

# ===== opcional: ordenar muestras por similitud =====
try:
    from scipy.spatial.distance import pdist
    from scipy.cluster.hierarchy import linkage, leaves_list
    SCIPY_OK = True
except Exception:
    SCIPY_OK = False

# =========================
# Parámetros
# =========================
N_TOP_PVAL       = 20       # semilla inicial por p-value (más significativo = menor p)
COVERAGE_TARGET  = 0.85     # cobertura mínima por muestra (0.80–0.95)
MAX_GENEROS      = 150      # tope de seguridad
OBJETIVO         = ["Cryptosporidium", "Entamoeba", "Blastocystis"]

TOP_LEYENDA      = 25       # leyenda compacta (Top + patógenos)
PATOGENOS_SET    = set(OBJETIVO)

# =========================
# 1) Copia y limpieza básica
# =========================
df = df_all_18SG.copy()
cols_muestras = [c for c in df.columns if c not in ["Género", "p-values"]]
if "SumaTotal" in cols_muestras:
    cols_muestras.remove("SumaTotal")
df[cols_muestras] = df[cols_muestras].apply(pd.to_numeric, errors="coerce").fillna(0)

# Renombres puntuales
df["Género"] = df["Género"].replace({
    "Frontonia-1": "Frontonia",
    "Frontonia-2": "Frontonia",
    "Colpodellida_CHR1_X": "Colpodellida",
    "Pseudoperkinsidae_XX": "Pseudoperkinsidae",
})

# Abreviar cualquier "Dino Group ..." -> "Dino-Group I/II/..."
def abreviar_dino(s: str) -> str:
    if not isinstance(s, str):
        return s
    t = s.replace("—", "-").replace("–", "-")
    m = re.search(r"(?i)dino[\s\-_]*group.*?([IVXLCDM]+)", t)
    if m:
        return f"Dino-Group {m.group(1).upper()}"
    if re.search(r"(?i)dino[\s\-_]*group", t):
        return "Dino-Group"
    return s

df["Género"] = df["Género"].apply(abreviar_dino)

def es_dino_group(nombre: str) -> bool:
    if not isinstance(nombre, str):
        return False
    return bool(re.search(r"(?i)^dino[\s\-_]*group", nombre))

# --- Abreviar variantes de Abeoformidae_* a "Abeoformidae Group" (2 palabras) ---
ABEO_PATTERN = r'(?i)\babeoformidae[\s_\-]*(group|grp)?(?:[\s_\-]*maip)?(?:[\s_\-]*[0-9xiv]+)?(?:[\s_\-]*[a-z])?\b.*'
df["Género"] = df["Género"].str.replace(ABEO_PATTERN, "Abeoformidae Group", regex=True)

# =========================
# 2) Agregación y p-values
# =========================
abund_por_gen = df.groupby("Género")[cols_muestras].sum()
# Para p-value por género usamos el mínimo (más significativo)
pval_por_gen = (df.groupby("Género")["p-values"].min()
                if "p-values" in df.columns else pd.Series(index=abund_por_gen.index, dtype=float))

# Presencia de patógenos (umbral muy bajo, en %)
MIN_REL_ABUND = 0.01
presentes = {g: (g in abund_por_gen.index and (abund_por_gen.loc[g] >= MIN_REL_ABUND).any())
             for g in OBJETIVO}

# =========================
# 3) Selección por p-value + Cobertura
# =========================
# 3.1 Semilla: Top por p-value ASC (menor = más significativo), excluyendo grupos "Dino-Group" si no los quieres
candidatos_p = pval_por_gen.dropna().sort_values(ascending=True).index.tolist()
candidatos_p = [g for g in candidatos_p if not es_dino_group(g)]  # comenta esta línea si deseas incluirlos

base_sel = list(dict.fromkeys(candidatos_p[:N_TOP_PVAL]))

# añade patógenos presentes (respetando su p-value si existe)
forzados_presentes = [g for g in OBJETIVO if presentes.get(g, False)]
forzados_presentes.sort(key=lambda x: pval_por_gen.get(x, np.inf))  # menor p primero
for g in forzados_presentes:
    if g not in base_sel:
        base_sel.append(g)

# 3.2 Expandir por cobertura (añadiendo por abundancia global)
tot_por_muestra = abund_por_gen.sum(axis=0).replace(0, np.nan)
rel = abund_por_gen.div(tot_por_muestra, axis=1).fillna(0.0)

def cobertura_actual(sel):
    if not sel:
        return pd.Series(0.0, index=rel.columns)
    return rel.loc[sel].sum(axis=0).clip(upper=1.0)

orden_abund = abund_por_gen.sum(axis=1).sort_values(ascending=False).index.tolist()
orden_abund = [g for g in orden_abund if not es_dino_group(g)]

seleccion = list(base_sel)
cov = cobertura_actual(seleccion)
i = 0
while (cov.min() < COVERAGE_TARGET) and (len(seleccion) < MAX_GENEROS):
    while i < len(orden_abund) and orden_abund[i] in seleccion:
        i += 1
    if i >= len(orden_abund):
        break
    seleccion.append(orden_abund[i])
    cov = cobertura_actual(seleccion)
    i += 1

# >>> Aquí insertas el bloque de muestras problemáticas <<<
muestras_obj = ["F4","F2","F3","C. virginica","F6","F9"]  # ajusta a tus casos
generos_extra = []
for m in muestras_obj:
    if m in abund_por_gen.columns:
        generos_extra += abund_por_gen[m].sort_values(ascending=False).head(3).index.tolist()
for g in generos_extra:
    if g not in seleccion:
        seleccion.append(g)

# 3.3 Orden final para la LEYENDA/COLores:
anadidos = [g for g in seleccion if g not in base_sel]
anadidos = abund_por_gen.loc[anadidos].sum(axis=1).sort_values(ascending=False).index.tolist()
seleccion_ordenada = base_sel + anadidos


# Diagnóstico
cov_diag = cobertura_actual(seleccion_ordenada)
print("Cobertura mínima:", float(cov_diag.min().round(3)))
print("Cobertura mediana:", float(cov_diag.median().round(3)))

# =========================
# 4) Normalización y “Otros”
# =========================
total_por_muestra = abund_por_gen.sum(axis=0).replace(0, np.nan)
full_norm = (abund_por_gen.T.div(total_por_muestra, axis=0).fillna(0) * 100)

top_norm = full_norm[seleccion_ordenada].copy()
mat_norm = top_norm.copy()
mat_norm["Otros"] = (100 - top_norm.sum(axis=1)).clip(lower=0)

# (opcional) ordenar muestras por similitud
if SCIPY_OK and mat_norm.shape[0] > 2:
    try:
        D = pdist(mat_norm.values, metric="braycurtis")
        Z = linkage(D, method="average")
        order_idx = leaves_list(Z)
        mat_norm = mat_norm.iloc[order_idx]
    except Exception:
        pass

# =========================
# 4.1) Reordenación manual del eje X
# =========================
def mover_despues(lista, item, despues_de):
    """Mueve 'item' justo después de 'despues_de' en la lista."""
    if item in lista and despues_de in lista:
        lista = [x for x in lista if x != item]
        idx = lista.index(despues_de)
        lista.insert(idx + 1, item)
    return lista

def mover_antes(lista, item, antes_de):
    """Mueve 'item' justo antes de 'antes_de' en la lista."""
    if item in lista and antes_de in lista:
        lista = [x for x in lista if x != item]
        idx = lista.index(antes_de)
        lista.insert(idx, item)
    return lista

orden_x = list(mat_norm.index)

# --- Caso 1: mover C. virginica después de F9 ---
aliases_cvir = ["C. virginica", "Crassostrea virginica", "C. virginica (promedio)"]
alias_cvir_presente = next((a for a in aliases_cvir if a in orden_x), None)

aliases_f9 = ["F6", "HecesP14", "F-9"]
alias_f9_presente = next((a for a in aliases_f9 if a in orden_x), None)

if alias_cvir_presente and alias_f9_presente:
    orden_x = mover_despues(orden_x, alias_cvir_presente, alias_f9_presente)

# --- Caso 2: mover RMW antes de GZW ---
alias_rmw = next((a for a in ["RMW"] if a in orden_x), None)
alias_gzw = next((a for a in ["GZW"] if a in orden_x), None)

if alias_rmw and alias_gzw:
    orden_x = mover_antes(orden_x, alias_rmw, alias_gzw)

# Aplicar nuevo orden si cambió
mat_norm = mat_norm.loc[orden_x]

# =========================
# 5) Plot “journal-ready”
# =========================
cols_plot = seleccion_ordenada + ["Otros"]

mpl.rcParams.update({
    "figure.dpi": 150, "savefig.dpi": 600, "savefig.bbox": "tight",
    "pdf.fonttype": 42, "ps.fonttype": 42,
    "axes.labelsize": 9, "xtick.labelsize": 8, "ytick.labelsize": 8,
    "legend.fontsize": 7, "axes.linewidth": 0.8,
    "font.family": "sans-serif", "font.sans-serif": ["Helvetica", "Arial", "DejaVu Sans"],
})

# Paleta lo-que-se-necesite
def make_palette(n_main: int):
    bancos = [plt.cm.tab20, plt.cm.tab20b, plt.cm.tab20c]  # 60 tonos
    colores = []
    for banco in bancos:
        if len(colores) >= n_main: break
        m = min(n_main - len(colores), 20)
        colores += [banco(i) for i in range(m)]
    i = 0
    while len(colores) < n_main:
        colores.append(plt.cm.tab20(i % 20)); i += 1
    return colores

n_main = max(0, len(cols_plot) - 1)
palette_main = make_palette(n_main)
palette = palette_main + ["#9E9E9E"]  # gris para "Otros"
color_map = {g: c for g, c in zip(cols_plot, palette)}

# Colores fijos para patógenos (si están en cols_plot)
fixed_colors = {
    "Cryptosporidium": "#800080",
    "Entamoeba": "#0084ff",
    "Blastocystis": "#ff7f0e",
}
for g, col in fixed_colors.items():
    if g in color_map:
        color_map[g] = col

DEFAULT_COLOR = "#808080"
for g in cols_plot:
    if g not in color_map:
        color_map[g] = DEFAULT_COLOR

def _sanitize(s): return s.replace("_", r"\_").replace("%", r"\%")
def italicize(lbl): return rf"$\it{{{_sanitize(lbl)}}}$" if lbl != "Otros" else lbl
def wrap_xticks(ax, width=10):
    ax.set_xticklabels([fill(t.get_text(), width) for t in ax.get_xticklabels()])

fig, ax = plt.subplots(figsize=(7.09, 3.5))
mat_norm[cols_plot].plot(
    kind="bar", stacked=True, ax=ax,
    color=[color_map.get(c, DEFAULT_COLOR) for c in cols_plot],
    width=0.92, edgecolor="none"
)

ax.set_ylabel("Abundancia relativa (%)")
ax.set_xlabel("Muestras")
ax.set_ylim(0, 100)
ax.yaxis.set_major_locator(mpl.ticker.MultipleLocator(20))
ax.yaxis.set_minor_locator(mpl.ticker.MultipleLocator(10))
ax.grid(axis="y", which="major", linestyle="--", linewidth=0.5, alpha=0.5)
ax.grid(axis="y", which="minor", linestyle=":", linewidth=0.35, alpha=0.3)
ax.spines["top"].set_visible(False); ax.spines["right"].set_visible(False)
ax.tick_params(axis="x", rotation=45)
wrap_xticks(ax, width=10)

# ===== Leyenda compacta (Top + patógenos), en 1–3 columnas =====
top_global = (abund_por_gen.loc[seleccion_ordenada]
              .sum(axis=1).sort_values(ascending=False)).index.tolist()

legend_keep = []
for g in top_global:
    if g not in legend_keep:
        legend_keep.append(g)
    if len(legend_keep) >= TOP_LEYENDA:
        break
for g in seleccion_ordenada:
    if g in PATOGENOS_SET and g not in legend_keep:
        legend_keep.append(g)
legend_keep.append("Otros")

handles_all, labels_all = ax.get_legend_handles_labels()
hmap = {}
for h, lab in zip(handles_all, labels_all):
    if lab not in hmap:
        hmap[lab] = h

handles_kept = [hmap[g] for g in legend_keep if g in hmap]
labels_kept  = [italicize(g) for g in legend_keep if g in hmap]

n_items = len([l for l in labels_kept if l != "Otros"])
ncol = 1 if n_items <= 20 else (2 if n_items <= 40 else 3)

ax.legend(
    handles_kept, labels_kept,
    title="Género",
    bbox_to_anchor=(1.02, 1.0), loc="upper left",
    frameon=False, ncol=ncol, title_fontsize=8,
    columnspacing=0.8, handlelength=1.2, handletextpad=0.4, borderaxespad=0.8
)

fig.tight_layout()
fig.savefig("fig18S_generos_top20_pvalue.tiff", dpi=600,
            pil_kwargs={"compression": "tiff_lzw"})
fig.savefig("fig18S_generos_top20_pvalue.pdf")
plt.show()


```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

# Configuración para que el PDF sea compatible con imprenta
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype']  = 42

# Cargar el archivo
archivo = "final.opti_mcc.groups.rarefaction"
df = pd.read_csv(archivo, sep='\t')

# Extraer columnas de rarefacción (las que comienzan con "0.03-")
muestras = [col for col in df.columns if col.startswith("0.03-")]

# Diccionario de reemplazo de nombres
renombres = {
    "AguaP1": "OCW",
    "AguaP2": "HZW",
    "AguaP3": "GZW",
    "AguaP4": "RMW",
    "AguaP5": "CZW",
    "CrassostreavirginicaG1": "O1",
    "CrassostreavirginicaG2": "O2",
    "CrassostreavirginicaG3": "O3",
    "CrassostreavirginicaG4": "O4",
    "CrassostreavirginicaG5": "O5",
    "CrassostreavirginicaG6": "O6",
    "CrassostreavirginicaG7": "O7",
    "CrassostreavirginicaG8": "O8",
    "CrassostreavirginicaG9": "O9",
    "CrassostreavirginicaG10": "O10",
    "HecesP6": "F1",
    "HecesP7": "F2",
    "HecesP8": "F3",
    "HecesP9": "F4",
    "HecesP10": "F5",
    "HecesP11": "F6",
    "HecesP12": "F7",
    "HecesP13": "F8",
    "HecesP14": "F9"
}

# Iniciar la gráfica
plt.figure(figsize=(14, 6))

# Graficar cada muestra con colores por grupo
for muestra in muestras:
    nombre_original = muestra.replace("0.03-", "")
    nombre_nuevo = renombres.get(nombre_original, nombre_original)

    # Definir colores por tipo de muestra
    if nombre_nuevo in ["OCW", "HZW", "GZW", "RMW", "CZW"]:
        color = "#1f77b4"  # azul agua
    elif nombre_nuevo.startswith("F"):
        if nombre_nuevo in ["F1", "F2", "F3", "F4"]:
            color = "#ff7f0e"  # naranja heces
        else:
            color = "#d62728"  # rojo heces
    elif nombre_nuevo.lower().startswith("crassostrea"):
        if nombre_nuevo in ["O1", "O2", "O3", "O4", "O5"]:
            color = "#2ca02c"  # verde ostión
    else:
        color = "#7f7f7f"  # gris por defecto

    plt.plot(df['numsampled'], df[muestra], label=nombre_nuevo, linewidth=2, color=color)

# Personalización
plt.xlabel('Número de secuencias muestreadas', fontsize=14)
plt.ylabel('OTUs observadas (Sobs)', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

# Leyenda en una sola columna, más grande
plt.legend(title='Muestra', bbox_to_anchor=(1.15, 1), loc='upper right',
           fontsize=12, title_fontsize=13, ncol=1, frameon=True)

# ==== ZOOM (ajusta los valores según tu rango de interés) ====
plt.xlim(0, 1000)   # eje X
plt.ylim(0, 20)     # eje Y

plt.tight_layout()
plt.grid(True)

# Exportar en PDF para publicación
plt.savefig("curvas_rarefaccion.pdf", format="pdf", dpi=300)

plt.show()
```
